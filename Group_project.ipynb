# Section 3: Feature Engineering

```python
# Section 3: Feature Engineering
from sklearn.preprocessing import StandardScaler
import pandas as pd

# Create a copy of the features to avoid modifying the original
X_clean = X.copy()

# 1. Handle Missing Values
# The UCI Heart Disease dataset (id=45) has a few missing values in the 'ca' and 'thal' columns.
# We will impute numerical missing values with the median.
X_clean.fillna(X_clean.median(), inplace=True)

# 2. Identify Numerical and Categorical Columns
# Based on the UCI Heart Disease dataset documentation
numerical_cols = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']
categorical_cols = ['sex', 'cp', 'fbs', 'restecg', 'exang', 'slope', 'ca', 'thal']

# 3. Normalize Numerical Data
# Standardizing numerical features to have a mean of 0 and variance of 1
scaler = StandardScaler()
X_clean[numerical_cols] = scaler.fit_transform(X_clean[numerical_cols])

# 4. One-Hot Encode Categorical Data
# Converting categorical variables into dummy/indicator variables
X_engineered = pd.get_dummies(X_clean, columns=categorical_cols, drop_first=True)

print(f"Feature Engineering Complete. Engineered dataset shape: {X_engineered.shape}")
display(X_engineered.head())
```
