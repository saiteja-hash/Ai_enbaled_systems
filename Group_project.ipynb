# Section 4: Initial Model Development

```python
# spliting the dataset into training and testing (75% training, 25% testing)
trainset, testset = train_test_split(heart_disease.data.df, test_size=0.25)

# initial recommendation model using svd
algo = SVD(random_state=42)

# training the model on the training set
algo.fit(trainset)

# making predictions on the test set
predictions = algo.test(testset)

# calculate performance metrics RMSE
rmse = accuracy.rmse(predictions)
mae = accuracy.mae(predictions)

print("RMSE:", rmse)
print("MAE:", mae)
```
